# Playgrounds

___

## Introduction - Rylie Loughran
I tried to find the best neighborhoods in Pittsburgh based on Playground Accessibility. Limited access to recreational outdoor spaces is a major barrier that leads to sedentary lifestyles and reduced community health and wellbeing. My goal is to identify neighborhoods where families and residents have the highest amount of public play spaces within their boundaries.

To measure availability, I used the City of Pittsburgh Playgrounds dataset to calculate the Total Playground Count for each neighborhood. This metric measures the total number of municipal playground facilities located within a neighborhood's geographic boundaries.

The core metric is the Total Playground Count:
- a high score means abundant public play spaces in the neighborhood, showing high accessibility and active outdoor lifestyle support (best)
- a low score means a lack of formal play infrastructure, showing low accessibility and underserved status (worst)

## Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("playgrounds.csv")

cleaned_df = df.dropna(subset=['neighborhood']).copy()

playground_counts = cleaned_df.groupby('neighborhood').size().reset_index(name='playground_count')

ranked_playgrounds = playground_counts.sort_values(by='playground_count', ascending=False).reset_index(drop=True)

print("\n Top 3 Neighborhoods by Playground Count:")
display(ranked_playgrounds.head(3))

best_hood = ranked_playgrounds.iloc[0]['neighborhood']
best_count = ranked_playgrounds.iloc[0]['playground_count']

print(f"\n ANSWER FOR PLAYGROUNDS METRIC: The best neighborhood for playgrounds is {best_hood} with {best_count} recorded playgrounds!")

ranked_all = playground_counts.sort_values(by='playground_count', ascending=True).reset_index(drop=True)

plt.bar(ranked_all['neighborhood'], ranked_all['playground_count'], color='skyblue', edgecolor='black', linewidth=0.5, alpha=0.85)

plt.xlabel('Neighborhood', fontsize=12, fontweight='bold')
plt.ylabel('Playground Count', fontsize=12, fontweight='bold')
plt.title('All Pittsburgh Neighborhoods Ranked by Playground Count', fontsize=14, fontweight='bold', pad=15)

plt.xticks(rotation=45, ha='right', fontsize = 6)

plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

## Conclusion
Playground access is the most accessible in Squirrel Hill South, Beechview, and Highland Park because they have multiple municipal parks. Residents of these neighborhoods have abundant options nearby to help encourage outdoor activity. By the metric of the number of playgrounds, Squirrel Hill South is the best.

# Pools

___

## Introduction - Rylie Loughran
I tried to find the best neighborhoods in Pittsburgh based on Public Pool Access. Limited access to municipal facilities for cooling off in the summer that double as swimming infrastructure during the hottest months is a problem that leads to reduced community recreation and lowered public health. 

To measure availability, I used the City of Pittsburgh Pools dataset to calculate the Total Pool Count for each neighborhood. This metric measures the total number of public pool facilities located within a neighborhood's geographic boundaries.

The core metric is the Total Pool Count:
- a high score means a high amount of public swimming facilities in the neighborhood, demonstrating high accessibility and summer active lifestyle support (best)
- a low score means a lack of formal infrastructure, which demonstrates low accessibility and government investment

## Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt

df2 = pd.read_csv("pools.csv")
cleaned_pools = df2.dropna(subset=['neighborhood']).copy()
pool_counts = cleaned_pools.groupby('neighborhood').size().reset_index(name='pool_count')
ranked_pools = pool_counts.sort_values(by='pool_count', ascending=False).reset_index(drop=True)
print("Top 3 Neighborhoods by Pool Count:")
display(ranked_pools.head(3))
best_pool_hood = ranked_pools.iloc[0]['neighborhood']
best_pool_count = ranked_pools.iloc[0]['pool_count']

print(f"\n ANSWER FOR POOLS METRIC: The best neighborhood for pools is {best_pool_hood} with {best_pool_count} recorded pools!")

ranked_pools_all = pool_counts.sort_values(by='pool_count', ascending=True).reset_index(drop=True)
plt.figure(figsize=(16, 6))
plt.bar(ranked_pools_all['neighborhood'], ranked_pools_all['pool_count'], color='dodgerblue', edgecolor='black', linewidth=0.5, alpha=0.85)

plt.xlabel('Neighborhood', fontsize=12, fontweight='bold')
plt.ylabel('Pool Count', fontsize=12, fontweight='bold')
plt.title('All Pittsburgh Neighborhoods Ranked by Pool Access', fontsize=14, fontweight='bold', pad=15)

# Rotate x-axis labels vertically so every neighborhood name is readable
plt.xticks(rotation=90, ha='right', fontsize=8)

plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

## Conclusion
The best neighborhoods by the metric of public pool access are Highland Park and South Side Flats because they have municipal pool facilities and aquatic centers. Residents of these neighborhoods have convenient options nearby to help encourage outdoor activity, particularly in the summer.
